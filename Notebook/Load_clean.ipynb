## Step 4: Data Cleaning

```python
# Keep only relevant loan statuses
# We only care about COMPLETED loans: Fully Paid or Charged Off
print('All loan status values before filtering:')
print(df_accepted['loan_status'].value_counts())
```

```python
# Keep only the two final statuses we care about
df_acc_clean = df_accepted[df_accepted['loan_status'].isin(['Fully Paid', 'Charged Off'])].copy()
print(f'Rows before filtering: {len(df_accepted)}')
print(f'Rows after  filtering: {len(df_acc_clean)}')
print(f'Rows removed         : {len(df_accepted) - len(df_acc_clean)}')
```

```python
# Keep only the two final statuses we care about
df_rej_clean = df_rejected.copy()
print(f'Rejected loans: {len(df_rej_clean)}')

```

```python
# Create binary target variable
# 0 = Fully Paid (good loan)
# 1 = Charged Off (default ==> Target to predict)

df_acc_clean['target'] = (df_acc_clean['loan_status'] == 'Charged Off').astype(int)

print('Target variable created:')
print(df_acc_clean['target'].value_counts())
print('0 = Fully Paid (no default)')
print('1 = Charged Off (default)')
```

```python
# extract term to number
if df_acc_clean['term'].dtype == object:
    df_acc_clean['term'] = df_acc_clean['term'].str.extract(r'(\d+)').astype(float)
    print('term column:')
    print(df_acc_clean['term'].value_counts())
```

```python
# We convert it to a number years

emp_map = {
    '< 1 year': 0, '1 year': 1, '2 years': 2, '3 years': 3,
    '4 years': 4, '5 years': 5, '6 years': 6, '7 years': 7,
    '8 years': 8, '9 years': 9, '10+ years': 10
}

if df_acc_clean['emp_length'].dtype == object:
    df_acc_clean['emp_length'] = df_acc_clean['emp_length'].map(emp_map)

print('emp_length after conversion:')
print(df_acc_clean['emp_length'].describe().round(2))
```

```python
# Fix emp_length for Rejected Loans DataFrame

emp_map = {
    '< 1 year': 0, '1 year': 1, '2 years': 2, '3 years': 3,
    '4 years': 4, '5 years': 5, '6 years': 6, '7 years': 7,
    '8 years': 8, '9 years': 9, '10+ years': 10
}

if df_rej_clean['Employment Length'].dtype == object:
    df_rej_clean['Employment Length'] = df_rej_clean['Employment Length'].map(emp_map)

print('Employment Length after conversion:')
print(df_rej_clean['Employment Length'].describe().round(2))
```

```python
#remove all outliers from the accepted loans dataset using the IQR method and then save the cleaned dataset
def remove_outliers_iqr(df, column):
    # Convert column to numeric
    df[column] = pd.to_numeric(df[column], errors='coerce')
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Convert numeric columns to numeric types before removing outliers
numeric_columns = ['loan_amnt', 'int_rate', 'installment', 'annual_inc', 'dti', 'fico_range_low', 'open_acc', 'pub_rec', 'revol_util']
for col in numeric_columns:
    if col in df_accepted.columns:
        df_accepted[col] = pd.to_numeric(df_accepted[col], errors='coerce')

# Remove outliers from the 'loan_amnt' column in the accepted loans dataset
df_no_outliers = remove_outliers_iqr(df_acc_clean.copy(), 'loan_amnt')
print(f'Accepted Loans after removing outliers: {df_no_outliers.shape[0]:,} rows')



```

```python
# remove outliers from the rejected loans dataset using the IQR method
df_rejected_no_outliers = remove_outliers_iqr(df_rej_clean.copy(), 'loan_amnt')
print(f'Rejected Loans after removing outliers: {df_rejected_no_outliers.shape[0]:} rows')
```

```python
# numeric and categorical columns
numeric_cols = df_acc_clean.select_dtypes(include="number").columns.tolist()
categorical_cols = df_acc_clean.select_dtypes(include="object").columns.tolist()

numeric_cols = [col for col in numeric_cols if col != "target"]

# Fill missing values for numeric columns with the median
for col in numeric_cols:
    missing_count = df_acc_clean[col].isnull().sum()
    if missing_count > 0:
        median_value = df_acc_clean[col].median()
        df_acc_clean[col].fillna(median_value, inplace=True)
        print(f"{col}: filled with median = {median_value:.2f} (missing: {missing_count})")
    else:
        print(f"{col}: no missing values")

# Fill missing values for categorical columns with the mode
for col in categorical_cols:
    # skip the loan_status column
    if col == "loan_status":
        continue

    missing_count = df_acc_clean[col].isnull().sum()
    if missing_count > 0:
        mode_values = df_acc_clean[col].mode()
        if not mode_values.empty:
            mode_value = mode_values.iloc[0]
            df_acc_clean[col].fillna(mode_value, inplace=True)
            print(f"{col}: filled with mode = {mode_value} (missing: {missing_count})")
    else:
        print(f"{col}: no missing values")

# check for remaining missing values
total_missing = df_acc_clean.isnull().sum().sum()
print("\nMissing Values Check")
print("-" * 30)
print(f"Total missing values remaining: {total_missing}")
```

```python
# Fill missing values for Rejected Loans
numeric_rej_cols     = df_rej_clean.select_dtypes(include='number').columns.tolist()
categorical_rej_cols = df_rej_clean.select_dtypes(include='object').columns.tolist()

# Fill numeric with median
for col in numeric_rej_cols:
    if df_rej_clean[col].isnull().sum() > 0:
        median_val = df_rej_clean[col].median()
        df_rej_clean[col] = df_rej_clean[col].fillna(median_val)
        print(f'  Filled {col} with median = {median_val:.2f}')

```

```python
# Fill categorical rejected data with mode 
for col in categorical_rej_cols:
    if df_rej_clean[col].isnull().sum() > 0:
        mode_series = df_rej_clean[col].mode()
        if not mode_series.empty:
            mode_value = mode_series.iloc[0]
            df_rej_clean[col] = df_rej_clean[col].fillna(mode_value)
            print(f"{col} Filled with mode: {mode_value}")
        else:
            print(f"{col} Skipped (all values are missing)")
            
# Verify: no missing values remain
total_rej_missing = df_rej_clean.isnull().sum().sum()
print(f'\nTotal missing values remaining With rejects: {total_rej_missing}')
```

```python
# Verify: no missing values remain
total_rej_missing = df_rej_clean.isnull().sum().sum()
print(f'\nTotal missing values remaining: {total_missing}')
```

```python
# Remove duplicate rows from Accepted Loans
before = len(df_acc_clean)
df_acc_clean = df_acc_clean.drop_duplicates()
after  = len(df_acc_clean)
print(f'Duplicates removed: {before - after}')
print(f'Final cleaned dataset: {df_acc_clean.shape[0]:,} rows × {df_acc_clean.shape[1]} columns')
```

```python
# Remove duplicate rows from Accepted Loans
before = df_acc_clean.shape[0]
df_acc_clean = df_acc_clean.drop_duplicates(keep='first', inplace=False)
after = df_acc_clean.shape[0]

# results
duplicates_removed = before - after
print(f"Duplicates removed: {duplicates_removed}")
print(f"Final cleaned dataset: {df_acc_clean.shape[0]:} rows × {df_acc_clean.shape[1]} columns")
```
